```python
def computepay(hours, rate):
    if hours > 40:
        pay = 40 * rate + (hours - 40) * rate * 1.5
    else:
        pay = hours * rate
    return pay


hours = float(input("Enter Hours: "))
rate = float(input("Enter Rate: "))
print("Pay:", computepay(hours, rate))
```

```python
def computegrade ():
  try:
      score = float(input("Enter score: "))
      # 先做守门员：小于 0.0 或 大于 1.0 统统判为 Bad score
      if score < 0.0 or score > 1.0:
          print("Bad score")
      elif score >= 0.9:
          print("A")
      elif score >= 0.8:
          print("B")
      elif score >= 0.7:
          print("C")
      elif score >= 0.6:
          print("D")
      else:
          print("F")
  except ValueError:
      print("Bad score")


computegrade ()
```
