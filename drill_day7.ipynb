```python
total = 0      # 记录所有数字的总和
count = 0      # 记录一共输入了多少个合法的数字

while True:
    line = input("Enter a number: ")
    
    # 1. 检查是否到达退出信号
    if line == 'done':
        break
    
    # 2. 尝试将输入转换为数字
    try:
        num = float(line)   # 或者 int(line)，用 float 可以兼容小数输入
    except ValueError:
        print("Invalid input")
        continue            # 遇到非法输入，跳过本次循环剩下的代码，直接开始下一次循环
    
    # 3. 如果成功通过了上面的转换，进行累加与计数
    total = total + num
    count = count + 1

# 4. 循环结束，计算平均值并输出
# 加一个 if count > 0 是好习惯，防止用户第一个输入就是 done 导致除以 0 报错
if count > 0:
    print(total, count, total / count)
else:
    print("No valid numbers entered.")
```

```python
maxNum = None   # 记录最大值
minNum = None   # 记录最小值

while True:
    line = input("Enter a number: ")
    
    # 1. 检查是否到达退出信号
    if line == 'done':
        break
    
    # 2. 尝试将输入转换为数字
    try:
        num = float(line)   # 或者 int(line)，用 float 可以兼容小数输入
    except ValueError:
        print("Invalid input")
        continue            # 遇到非法输入，跳过本次循环剩下的代码，直接开始下一次循环
    
    # 3. 如果成功通过了上面的转换，进行最大值和最小值的更新
    if maxNum is None or num > maxNum:
        maxNum = num
    if minNum is None or num < minNum:
        minNum = num

# 4. 循环结束，输出最大值和最小值
if maxNum is not None and minNum is not None:
    print("Maximum:", maxNum)
    print("Minimum:", minNum)
else:
    print("No valid numbers entered.")
```
